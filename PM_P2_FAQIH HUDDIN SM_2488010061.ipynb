# **Refleksi 1**

supervised menggunakan data yang sudah memiliki label atau jawaban. Model belajar dari data tersebut untuk memprediksi hasil pada data baru (Data BERLABEL. Belajar memetakan fitur ke jawaban)

unsupervised menggunakan data yang tidak memiliki label atau jawaban. Model mencari pola atau kelompok tertentu dari data secara mandiri (Data TANPA LABEL. Menemukan struktur tersembunyi)

reinforcement dengan cara mencoba suatu tindakan dan mendapatkan reward (hadiah) atau punishment (hukuman) dari hasil tindakan tersebut (Belajar dari INTERAKSI via reward & punishment)
